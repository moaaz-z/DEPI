1.Extract Data

```python
import requests
import selenium

cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
list_cities=[]

url = "https://api.openweathermap.org/data/2.5/weather"

for city in cities:
    params = {
        "q": city,
        "appid": "6a13e61e853763292101962a026aa3e8",  
        "units": "metric"
    }

    response = requests.get(url, params=params)

    
    data = response.json()
    city_weather = {
            "city": city,
            "country": data["sys"]["country"],
            "temperature": data["main"]["temp"],
            "pressure": data["main"]["pressure"],
            "wind speed": data["wind"]["speed"],
            'humidity':data["main"]["humidity"]

        }
    list_cities.append(city_weather)
      


list_cities

  



```

2.Transform

```python
import pandas as pd
data = pd.DataFrame(list_cities)
data
```

```python
def fun (city):
    if city == 'Alexandria':
        return 'Upper Cairo'
    elif city == 'Suez':
        return 'Canal'
    elif city == 'Ismailia':
        return 'Canal'
    elif city == 'Port Said':
        return 'Canal'
    elif city == 'Cairo':
        return 'Greater Cairo'
    elif city == 'Giza':
        return 'Greater Cairo'
    elif city == 'Mansoura':
        return 'Delta'
    elif city == 'Tanta':
        return 'Delta'
    elif city == 'Luxor':
        return 'Lower Egypt'
    elif city == 'Aswan':
        return 'Lower Egypt'
    
```

```python
data['Region']= data['city'].apply(fun)
data
```

```python
def tempcat(temp):
    if temp < 10:
        return 'Cold'
    elif temp >= 10 and temp < 20:
        return 'Mild'
    elif temp >= 20 and temp < 30:
        return 'Warm'
    else:
        return 'Hot'
```

```python
data['TemperatureCategory']=data['temperature'].apply(tempcat)
data
```

```python
def HumidityCategory(humidity):
    if humidity < 30:
        return 'Low'
    elif humidity >= 30 and humidity < 60:
        return 'Medium'
    else:
        return 'High'
```

```python
data['HumidityCategory']=data['humidity'].apply(HumidityCategory)
data
```
