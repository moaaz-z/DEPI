```python
import requests
import selenium

cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
list_cities=[]

url = "https://api.openweathermap.org/data/2.5/weather"

for city in cities:
    params = {
        "q": city,
        "appid": "6a13e61e853763292101962a026aa3e8",  
        "units": "metric"
    }

    response = requests.get(url, params=params)

    
    data = response.json()
    city_weather = {
            "city": city,
            "temperature": data["main"]["temp"],
            "description": data["weather"][0]["description"]
        }
    list_cities.append(city_weather)
      


list_cities

  



```
