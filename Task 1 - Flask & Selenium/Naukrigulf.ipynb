```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time

options = Options()
options.binary_location = "/usr/bin/brave-browser"
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option("useAutomationExtension", False)
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(options=options)

driver.get("https://www.naukrigulf.com")
time.sleep(3)

driver.find_element(
    By.XPATH,
    "//div[@class='qsb-proxy']"
).click()

time.sleep(2)

driver.find_element(
    By.CSS_SELECTOR,
    "#qsbKey"
).send_keys("Data Engineer")

driver.find_element(
    By.CSS_SELECTOR,
    "#qsbLoc"
).send_keys("Egypt")

driver.find_element(
    By.CSS_SELECTOR,
    "input[value='Search Jobs']"
).click()

time.sleep(4)


# Get only actual job links
job_elements = driver.find_elements(
    By.CSS_SELECTOR,
    "a[href*='jid-']"
)

links = []

for job in job_elements:

    link = job.get_attribute("href")

    if link and link not in links:
        links.append(link)


print("Jobs found:", len(links))


all_jobs = {}


for link in links:

    driver.get(link)
    time.sleep(2)

    try:

        # Job title
        title = driver.find_element(
            By.TAG_NAME,
            "h1"
        ).text.strip()


        # Get all page text
        text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        lines = [
            line.strip()
            for line in text.split("\n")
            if line.strip()
        ]


        # Company
        company = "N/A"

        if title in lines:
            index = lines.index(title)

            if index + 1 < len(lines):
                company = lines[index + 1]


        # Experience
        experience = "N/A"

        if "Experience" in lines:
            index = lines.index("Experience")

            if index + 1 < len(lines):
                experience = lines[index + 1]


        # Location
        location = "N/A"

        if "Job Location" in lines:
            index = lines.index("Job Location")

            if index + 1 < len(lines):
                location = lines[index + 1]


        # Full Job Description
        description = "N/A"

        if "Job Description" in lines:

            start = lines.index("Job Description") + 1

            if "Company Industry" in lines:
                end = lines.index("Company Industry")
            else:
                end = len(lines)

            description = " ".join(
                lines[start:end]
            )


        all_jobs[link] = {
            "title": title,
            "company": company,
            "location": location,
            "experience": experience,
            "description": description
        }


        print("\n----------------------")
        print("Title:", title)
        print("Company:", company)
        print("Location:", location)
        print("Experience:", experience)


    except Exception as e:
        print("Error:", link, e)


print("\n========================")
print("Total unique jobs:", len(all_jobs))
print("========================")


for link, job in all_jobs.items():

    print("\nTitle:", job["title"])
    print("Company:", job["company"])
    print("Location:", job["location"])
    print("Experience:", job["experience"])
    print("Description:", job["description"])
    print("URL:", link)
    print("------------------------")


input("Press Enter to close...")

driver.quit()
```
