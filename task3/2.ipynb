```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace("%", "percent", regex=False)
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df.columns.tolist())
```

```python
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()
```

```python
df["payment"] = df["payment"].replace({
    "E-wallet": "Ewallet",
    "credit card": "Credit card",
    "cash": "Cash"
})

df["payment"].value_counts()
```

```python
df["sales"] = (
    df["sales"]
    .astype("string")
    .str.replace("EGP", "", regex=False)
    .str.strip()
)

df["sales"] = pd.to_numeric(df["sales"], errors="coerce")

df["sales"].head()
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

df["date"].head()
```

```python
print(df.isnull().sum())
```

```python
df["unit_price"] = df["unit_price"].fillna(
    df["unit_price"].median()
)

df["sales"] = df["sales"].fillna(
    df["sales"].median()
)
```

```python
df["date"] = df["date"].ffill().bfill()
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
df = df.drop(columns=["invoice_id"])

df.head()
```

```python
df = df.sort_values(
    by=["date", "time"]
).reset_index(drop=True)

df.head()
```

```python
df["day"] = df["date"].dt.day
df["month"] = df["date"].dt.month
df["year"] = df["date"].dt.year

df[["date", "day", "month", "year"]].head()
```

```python
print("Shape:", df.shape)

print("\nNull values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
total_revenue = df["sales"].sum()

print("Total Revenue:", total_revenue)
```

```python
revenue_by_city = (
    df.groupby("city")["sales"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_city)

print("\nCity with highest revenue:")
print(revenue_by_city.idxmax())

print("Highest revenue:")
print(revenue_by_city.max())
```

```python
profit_by_branch = (
    df.groupby("branch")["gross_income"]
    .sum()
    .sort_values(ascending=False)
)

print(profit_by_branch)

print("\nMost profitable branch:")
print(profit_by_branch.idxmax())

print("Profit:")
print(profit_by_branch.max())
```

```python
category_analysis = (
    df.groupby("product_line")
    .agg(
        total_revenue=("sales", "sum"),
        total_profit=("gross_income", "sum")
    )
    .sort_values(
        by="total_profit",
        ascending=False
    )
)

print(category_analysis)

print("\nTop-performing category:")
print(category_analysis.index[0])
```

```python
spending_by_customer_type = (
    df.groupby("customer_type")["sales"]
    .sum()
    .sort_values(ascending=False)
)

print(spending_by_customer_type)

print("\nCustomer type that spends more:")
print(spending_by_customer_type.idxmax())
```

```python
transactions_by_payment = (
    df["payment"]
    .value_counts()
)

print(transactions_by_payment)

print("\nMost popular payment method:")
print(transactions_by_payment.idxmax())
```

```python
average_transaction_value = df["sales"].mean()

print(
    "Average transaction value:",
    average_transaction_value
)
```

```python
revenue_by_day = (
    df.groupby("day")["sales"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_day)

print("\nDay with highest sales:")
print(revenue_by_day.idxmax())

print("Revenue:")
print(revenue_by_day.max())
```

```python
revenue_by_month = (
    df.groupby("month")["sales"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_month)

print("\nMonth with highest sales:")
print(revenue_by_month.idxmax())

print("Revenue:")
print(revenue_by_month.max())
```
